```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None
       
    def add_node(self, value):
        if value < self.data:
            if self.left is None:
                self.left = Node(value)
            else:
                self.left.add_node(value)
        elif value > self.data:
            if self.right is None:
                self.right = Node(value)
            else:
                self.right.add_node(value)
    
    def NLR(self):
        print(self.data, end = " ")
        if self.left:
            self.left.NLR()
        if self.right:
            self.right.NLR()

    def LNR(self):
        
        if self.left:
            self.left.LNR()
        print(self.data, end = " ")
        if self.right:
            self.right.LNR()

    def search(self, value):
        if value == self.data:
            return True
        
        if self.left:
            self.left.search(value)

        if self.right:
            self.right.search(value)

        return False
    
    def count(self):
        count = 1
        if self.left:
            count += self.left.count()
        if self.right:
            count += self.right.count()
        return count
```

```python
lst = [5, 8, 1, 3, 2]
root = Node(lst[0])
for i in lst[1:]:
    #print(i)
    root.add_node(i)
```

```python
root.NLR()
print("\n")
root.LNR()
```

```python
root.search(5)
```
