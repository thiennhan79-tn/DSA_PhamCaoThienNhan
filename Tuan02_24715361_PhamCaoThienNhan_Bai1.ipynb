```python
class Student:
    def __init__(self, id, name, age, gpa):
        self.id = id
        self.name = name
        self.age = age
        self.gpa = gpa

    def InThongTin(self):
        return f"id:{self.id}, name:{self.name}, age:{self.age}, gpa:{self.gpa}"
```

```python
class RegularStudent(Student):
    def __init__(self, id, name, age, gpa, ):
        super().__init__(id, name, age, gpa)

    def InThongTin(self):
        return "Regular: " + super().InThongTin()
```

```python
class InternationalStudent(Student):
    def __init__(self, id, name, age, gpa, country, language_score):
        super().__init__(id, name, age, gpa)
        self.country = country
        self.language_score = language_score

    def InThongTin(self):
        return f"International: {super().InThongTin()} - Country:{self.country} - Language:{self.language_score}"
```

```python
class GraduateStudent(Student):
    def __init__(self, id, name, age, gpa, research_area, thesis_title):
        super().__init__(id, name, age, gpa)
        self.research_area = research_area
        self.thesis_title = thesis_title

    def InThongTin(self):
        return f"Graduate: {super().InThongTin()} - Research Area:{self.research_area} - Thesis Title:{self.thesis_title}"
```

```python
from abc import ABC, abstractmethod

class BaseSV(ABC):
    
    @abstractmethod
    def isEmpty(self):
        pass

    @abstractmethod
    def push(self, sv):
        pass

    @abstractmethod
    def pop(self):
        pass 

    @abstractmethod
    def show(self):
        pass
```

```python
class Stack:
    def __init__(self):
        self.data = []

    def isEmpty(self):
        return len(self.data) == 0

    def push(self, item):
        self.data.append(item)

    def pop(self):
        return None if self.isEmpty() else self.data.pop()

    def show(self):
        return self.data
```

```python
class Queue(BaseSV):
    def __init__(self):
        self.data = []

    def isEmpty(self):
        return len(self.data) == 0
    
    def push(self, item):
        self.data.append(item)

    def pop(self):
        if self.isEmpty():
            return None
        return self.data.pop(0)

    def show(self):
        return self.data
```

```python
class StudentManagement:
    def __init__(self):
        self.stack = Stack()

    def add_student(self, student):
        self.stack.push(student)

    def search_by_id(self, id):
        for s in self.stack.data:
            if s.id == id:
                return s
        return None
    
    def update_student(self, id, name=None, age=None, gpa=None):
        s = self.search_by_id(id)
        if s:
            if name: s.name = name
            if age: s.age = age
            if gpa: s.gpa = gpa
            return "Updated successfully"
        return "Student not found"

    def display_students(self):
        if self.stack.isEmpty():
            return "Danh sach sinh vien rong"
        return "\n".join(sv.InThongTin() for sv in self.stack.show())

    def find_oldest_student(self):
        if self.stack.isEmpty():
            return None
        return max(self.stack.data, key=lambda s: s.age)

    def find_highest_gpa(self):
        max_gpa = max(s.gpa for s in self.stack.data)
        return [s for s in self.stack.data if s.gpa == max_gpa]

    def bubble_sort_by_gpa(self):
        n = len(self.stack.data)
        for i in range (n):
            for j in range (0, n-1-i):
                if self.stack.data[j].gpa > self.stack.data[j+1].gpa:
                    self.stack.data[j], self.stack.data[j+1] = self.stack.data[j+1], self.stack.data[j]
    
    def selection_sort_by_id(self):
        n = len(self.stack.data)
        for i in range(n):
            min_idx = i
            for j in range(i + 1, n):
                if self.stack.data[j].id < self.stack.data[min_idx].id:
                    min_idx = j
            self.stack.data[i], self.stack.data[min_idx] = self.stack.data[min_idx], self.stack.data[i]
```

```python
def menu():
    sm = StudentManagement()

    while True:
        print("\n===== STUDENT MANAGEMENT SYSTEM =====")
        print("1. Add student")
        print("2. Display students")
        print("3. Find oldest student")
        print("4. Find student(s) with highest GPA")
        print("5. Sort students by GPA (Bubble Sort)")
        print("6. Sort students by GPA (Selection Sort)")
        print("0. Exit")

        choice = input("Choose an option: ")

        if choice == "1":
            print("\n--- Add Student ---")
            loai = input("Type (R: Regular, I: International, G: Graduate): ").upper()

            id = input("ID: ")
            name = input("Name: ")
            age = int(input("Age: "))
            gpa = float(input("GPA: "))

            if loai == "R":
                sv = RegularStudent(id, name, age, gpa)

            elif loai == "I":
                country = input("Country: ")
                language = float(input("Language score: "))
                sv = InternationalStudent(id, name, age, gpa, country, language)

            elif loai == "G":
                research = input("Research area: ")
                thesis = input("Thesis title: ")
                sv = GraduateStudent(id, name, age, gpa, research, thesis)

            else:
                print("Invalid student type")
                continue

            sm.add_student(sv)
            print("Student added successfully")

        elif choice == "2":
            print("\n--- Student List ---")
            print(sm.display_students())

        elif choice == "3":
            sv = sm.find_oldest_student()
            if sv:
                print("\nOldest student:")
                print(sv.InThongTin())
            else:
                print("No students")

        elif choice == "4":
            print("\nStudent(s) with highest GPA:")
            for sv in sm.find_highest_gpa():
                print(sv.InThongTin())

        elif choice == "5":
            sm.bubble_sort_by_gpa()
            print("Sorted by GPA")
            print(sm.display_students())

        elif choice == "6":
            sm.selection_sort_by_id()
            print("Sorted by GPA (Selection Sort)")
            print(sm.display_students())

        elif choice == "0":
            print("Exit program")
            break

        else:
            print("Invalid choice, try again")
```

```python
menu()
```
