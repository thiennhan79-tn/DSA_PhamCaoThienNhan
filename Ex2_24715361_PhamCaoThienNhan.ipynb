```python
class StudentNode:
    def __init__(self, student_id):
        self.student_id = student_id
        self.next = None
```

```python
class Course:
    def __init__(self, code, title, credits, instructor, capacity):
        self.code = code
        self.title = title
        self.credits = credits
        self.instructor = instructor
        self.capacity = capacity
        self.enrollment = 0

        self.students_head = None   
        self.next = None            
```

```python
class CourseList:
    def __init__(self):
        self.head = None
    
    def is_empty(self):
        return self.head is None

    def count_courses(self):
        count = 0
        temp = self.head

        while temp:
            count += 1
            temp = temp.next

        return count

    def add_first(self, code, title, credits, instructor, capacity):
        new_course = Course(code, title, credits, instructor, capacity)

        new_course.next = self.head
        self.head = new_course

    def add_last(self, code, title, credits, instructor, capacity):
        new_course = Course(code, title, credits, instructor, capacity)

        if self.is_empty():
            self.head = new_course
            return

        temp = self.head
        while temp.next:
            temp = temp.next

        temp.next = new_course
    
    def add_after(self, target_code, code, title, credits, instructor, capacity):
        temp = self.head

        while temp:
            if temp.code == target_code:
                new_course = Course(code, title, credits, instructor, capacity)
                new_course.next = temp.next
                temp.next = new_course
                return

            temp = temp.next

        print("Không tìm thấy môn học")

    def display(self):
        if self.is_empty():
            print("Danh sách rỗng")
            return

        temp = self.head

        while temp:
            print(temp.code, temp.title, temp.credits,
                  temp.instructor, temp.enrollment, "/", temp.capacity)
            temp = temp.next

    def find_course(self, code):
        temp = self.head

        while temp:
            if temp.code == code:
                return temp

            temp = temp.next

        return None

    def update_course(self, code, title, credits, instructor, capacity):
        course = self.find_course(code)

        if course:
            course.title = title
            course.credits = credits
            course.instructor = instructor
            course.capacity = capacity
        else:
            print("Không tìm thấy môn học")

    def remove_first(self):
        if self.is_empty():
            return

        self.head = self.head.next

    def remove_last(self):
        if self.is_empty():
            return

        if self.head.next is None:
            self.head = None
            return

        temp = self.head
        while temp.next.next:
            temp = temp.next

        temp.next = None

    def remove_by_code(self, code):
        if self.is_empty():
            return

        if self.head.code == code:
            self.head = self.head.next
            return

        prev = self.head
        temp = self.head.next

        while temp:
            if temp.code == code:
                prev.next = temp.next
                return

            prev = temp
            temp = temp.next

        print("Không tìm thấy môn học")

    def register_student(self, course_code, student_id):
        course = self.find_course(course_code)

        if course is None:
            print("Không tìm thấy môn học")
            return

        if course.enrollment >= course.capacity:
            print("Môn học đã đầy")
            return

        new_student = StudentNode(student_id)

        new_student.next = course.students_head
        course.students_head = new_student

        course.enrollment += 1

    def drop_student(self, course_code, student_id):
        course = self.find_course(course_code)

        if course is None:
            return

        prev = None
        temp = course.students_head

        while temp:
            if temp.student_id == student_id:
                if prev:
                    prev.next = temp.next
                else:
                    course.students_head = temp.next

                course.enrollment -= 1
                return

            prev = temp
            temp = temp.next

    def list_students(self, course_code):
        course = self.find_course(course_code)

        if course is None:
            return

        temp = course.students_head

        while temp:
            print(temp.student_id)
            temp = temp.next

    def enrollment_stats(self):
        if self.is_empty():
            return

        highest = lowest = self.head
        temp = self.head

        while temp:
            if temp.enrollment > highest.enrollment:
                highest = temp

            if temp.enrollment < lowest.enrollment:
                lowest = temp

            temp = temp.next

        print("Môn nhiều SV nhất:", highest.title)
        print("Môn ít SV nhất:", lowest.title)

    def sort_by_enrollment(self):
        arr = []
        temp = self.head

        while temp:
            arr.append(temp)
            temp = temp.next

        arr.sort(key=lambda x: x.enrollment)

        self.head = arr[0]
        temp = self.head

        for i in range(1, len(arr)):
            temp.next = arr[i]
            temp = temp.next

        temp.next = None

    def sort_by_title(self):
        arr = []
        temp = self.head

        while temp:
            arr.append(temp)
            temp = temp.next

        arr.sort(key=lambda x: x.title)

        self.head = arr[0]
        temp = self.head

        for i in range(1, len(arr)):
            temp.next = arr[i]
            temp = temp.next

        temp.next = None

    def save_to_file(self, filename):
        with open(filename, "w") as f:
            temp = self.head

            while temp:
                f.write(f"{temp.code},{temp.title},{temp.credits},{temp.instructor},{temp.capacity},{temp.enrollment}\n")
                temp = temp.next

    def load_from_file(self, filename):
        with open(filename, "r") as f:
            for line in f:
                code, title, credits, instructor, capacity, enrollment = line.strip().split(",")

                course = Course(code, title, int(credits), instructor, int(capacity))
                course.enrollment = int(enrollment)

                self.add_last(code, title, int(credits), instructor, int(capacity))
```

```python
courses = CourseList()

courses.add_last("IT101", "Python Programming", 3, "Dr. An", 30)
courses.add_last("IT102", "Data Structures", 4, "Dr. Binh", 40)

courses.register_student("IT101", "SV01")
courses.register_student("IT101", "SV02")

courses.display()

courses.list_students("IT101")
```
