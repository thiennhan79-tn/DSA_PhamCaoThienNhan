```python
class Node:
    def __init__(self, data = None):
        self.data = data
        self.left = None
        self.right = None

    def add_node(self, value):
        if value < self.data:
            if self.left is None:
                self.left = Node(value)
            else:
                self.left.add_node(value)
        elif value > self.data:
            if self.right is None:
                self.right = Node(value)
            else:
                self.right.add_node(value)
    
    def search(self, value):
        if value == self.data:
            return True
        if self.left:
            self.left.search(value)
        if self.right:
            self.right.search(value)
        return False
    
    def count_node(self):
        count = 1
        if self.left:
            count += self.left.count_node()
        if self.right:
            count += self.right.count_node()
        return count

    def NLR(self):
        print(self.data, end = " ")
        if self.left:
            self.left.NLR()
        if self.right:
            self.right.NLR()
    
    def RLN(self):
        if self.right:
            self.right.RLN()
        if self.left:
            self.left.RLN()
        print(self.data, end = " ")
    
    def LRN(self):
        if self.left:
            self.left.LRN()
        if self.right:
            self.right.LRN()
        print(self.data, end = " ")
    
    def count_node_la(self):
        if self.left is None and self.right is None:
            return 1
        count = 0
        if self.left:
            count += self.left.count_node_la()
        if self.right:
            count += self.right.count_node_la()
        return count

    def count_node_1_con(self):
        if (self.left and not self.right) or (not self.left and self.right):
            return 1
        count = 0
        if self.left:
            count += self.left.count_node_1_con()
        if self.right:
            count += self.right.count_node_1_con()
        return count
    
    def count_node_2_con(self):
        if self.left and self.right:
            return 1
        count = 0
        if self.left:
            count += self.left.count_node_2_con()
        if self.right:
            count += self.right.count_node_2_con()
        return count
    
    def sum_node(self):
        tong = self.data
        if self.left:
            tong += self.left.sum_node()
        if self.right:
            tong += self.right.sum_node()
        return tong
    
    def find_max(self):
        if self.right:
            return self.right.find_max()
        return self.data

    def find_min(self):
        if self.left:
            return self.left.find_min()
        return self.data
    
    def in_chan(self):
        if self.data % 2 == 0:
            print(self.data, end = " ")
        if self.left:
            self.left.in_chan()
        if self.right:
            self.right.in_chan()

    def in_le(self):
        if self.data % 2 != 0:
            print(self.data, end = " ")
        if self.left:
            self.left.in_le()
        if self.right:
            self.right.in_le()

    def delete(self, value):
        if value < self.data:
            if self.left:
                self.left = self.left.delete(value)
        elif value > self.data:
            if self.right:
                self.right = self.right.delete(value)
        else:
            if self.left is None and self.right is None:
                return None

            if self.left is None:
                return self.right
            if self.right is None:
                return self.left

            min_val = self.right.find_min()
            self.data = min_val
            self.right = self.right.delete(min_val)
        
        return self
    
    def LNR(self):
        if self.left:
            self.left.LRN()
        print(self.data, end = " ")
        if self.right:
            self.right.LRN()
        

    def is_prime(self, n):
            if n < 2:
                return False
            for i in range(2, int(n**0.5) + 1):
                if n % i == 0:
                    return False
            return True

    def print_prime(self):
        if self.is_prime(self.data):
            print(self.data, end=" ")
        if self.left:
            self.left.print_prime()
        if self.right:
            self.right.print_prime()


    def read_file(path):
        with open(path, "r") as f:
            arr = list(map(int, f.read().split()))
        root = Node(arr[0])
        for x in arr[1:]:
            root.add_node(x)
        return root


    def write_file(root, path):
        with open(path, "w") as f:
            def inorder(node):
                if node:
                    inorder(node.left)
                    f.write(str(node.data) + " ")
                    inorder(node.right)
            inorder(root)
```
