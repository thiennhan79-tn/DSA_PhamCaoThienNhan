```python
class Node:
    def __init__(self, data):
        self.data = data
        self.children = []
        
```

```python
class Tree:
    def __init__(self):
        self.root = None

    def add_node(self, parent, data):
        node = Node(data)
        if parent is None:
            self.root = node
        else:
            parent.children.append(node)
        return node
    
    def NLR(self, node = None):
        if node is None:
            node = self.root
        if node:
            print(node.data, end = " ")
            for c in node.children:
                self.NLR(c)

    def LRN(self, node=None):
        if node is None:
            node = self.root
        if node:
            for c in node.children:
                self.LRN(c)
            print(node.data, end=" ")
```

```python
t = Tree()
nodeR = t.add_node(None, "R")

nodeA = t.add_node(nodeR, "A")
nodeC = t.add_node(nodeA, "C")
nodeD = t.add_node(nodeA, "D")


nodeB = t.add_node(nodeR, "B")
nodeE = t.add_node(nodeB, "E")
nodeF = t.add_node(nodeB, "F")
nodeG = t.add_node(nodeF, "G")

```
