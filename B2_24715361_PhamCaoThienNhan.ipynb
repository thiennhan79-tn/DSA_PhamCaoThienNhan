**Lưu ý:** Tên thuộc tính dùng **tiếng Việt không dấu**: `ma`, `ten`, `tac_gia`, `gia`, `nam_xb`, ...

## 1) class

- **Class `Sach`:** Lưu thông tin `ma`, `ten`, `tac_gia`, `gia`, `nam_xb`.
  - Kiểm tra dữ liệu: `gia >= 0`, `nam_xb > 0` (dùng getter/setter hoặc validate trong `__init__`).

- **Kế thừa:** Ít nhất 2 lớp con từ `Sach`:
  1. `SachGiaoKhoa`: thêm `lop`, `mon`.
  2. `SachThamKhao`: thêm `chuyen_nganh`, `cap_do`.

- **Yêu cầu thêm hàm thời gian xuất bản:**
  - Trong lớp `Sach` tạo hàm `thoi_gian_xuat_ban()` để lớp con kế thừa/ghi đè.
  - Quy ước:
    - `SachGiaoKhoa`: xuất bản tháng 9/2026.
    - `SachThamKhao`: xuất bản tháng 3, 4, 5/2026.

## 2) Stack/Queue + Abstract

- Tạo `StackSach` hoặc `QueueSach` để quản lý sách.
- Dùng `abstractmethod` cho các hàm chính trong stack/queue, ví dụ:
  - `push`, `pop`, `peek`, `is_empty` (cho stack)
  - `enqueue`, `dequeue`, `front`, `is_empty` (cho queue)

## 3) BookManagement Class

Quản lý danh sách sách với các chức năng:
1. **Thêm sách:** nhập theo số lượng `n` (chọn `Sach` / `SachGiaoKhoa` / `SachThamKhao`), hoặc thêm từ file cho **một loại** (CSV/JSON).
2. Xóa sách theo `ma`.
3. Hiển thị danh sách sách.
4. Tìm sách theo `ma`.
5. Cập nhật thông tin sách:
   - Basic: cập nhật `ten`, `tac_gia`, `gia`.
   - Nâng cao: nhập `ma` -> kiểm tra loại sách -> cho cập nhật đúng thuộc tính của loại đó.
6. Tìm kiếm linh hoạt:
   - Tìm theo `gia` (chính xác hoặc theo khoảng).
   - Hoặc tìm theo `nam_xb`.
   - Hoặc tìm theo từ khóa `ten`.

## 4) Sắp xếp (đủ 5 thuật toán)

Mỗi thuật toán một tiêu chí:
- **Bubble Sort** — theo `gia`
- **Selection Sort** — theo `ma`
- **Insertion Sort** — theo `ten`
- **Merge Sort** — theo `nam_xb`
- **Quick Sort** — theo `gia` hoặc `ma` (khác Bubble/Selection)

## 5) File I/O

- Đổ dữ liệu sách từ file vào danh sách (CSV/JSON).
- Lưu danh sách hiện tại ra file (CSV/JSON).

## 6) Menu  

Tạo hàm `menu()` chạy vòng lặp `while True`, cho người dùng chọn chức năng bằng số:

7. Sắp xếp Bubble theo giá
8. Sắp xếp Selection theo mã
9. Sắp xếp Insertion theo tên
10. Sắp xếp Merge theo năm xuất bản
11. Sắp xếp Quick theo tiêu chí còn lại
12. Đọc dữ liệu từ file
13. Lưu dữ liệu ra file

**Yêu cầu menu:**
- Có bắt lỗi nhập sai lựa chọn (`try/except`).
- Sau mỗi thao tác phải in kết quả hoặc thông báo thành công/thất bại.
- Chỉ thoát khi người dùng chọn `0`.




```python
from abc import ABC, abstractmethod
class Book(ABC):
    def __init__(self, ma, hoten, tac_gia, gia, nam_xb):
        self.ma = ma
        self.hoten = hoten
        self.tac_gia = tac_gia
        self.gia = gia
        self.nam_xb = nam_xb

    @abstractmethod
    def ThoiGianXuatBan(self):
        pass

    def __str__(self):
        return f"Ma: {self.ma}, Ho ten: {self.hoten}, Tac gia: {self.tac_gia}, Gia: {self.gia}, Nam XB: {self.nam_xb}"

        
```

```python
class SachGiaoKhoa(Book):
    def __init__(self, ma, hoten, tac_gia, gia, nam_xb, lop, mon):
        super().__init__(ma, hoten, tac_gia, gia, nam_xb)
        self.lop = lop
        self.mon = mon

    def ThoiGianXuatBan(self):
        return "Xuat ban thang 9/26"

    def __str__(self):
        return f"{super.__str__()}, Lop: {self.lop}, Mon: {self.mon}"
```

```python
class SachThamKhao(Book):
    def __init__(self, ma, hoten, tac_gia, gia, nam_xb, chuyen_nganh, cap_do):
        super().__init__(ma, hoten, tac_gia, gia, nam_xb)
        self.chuyen_nganh = chuyen_nganh
        self.cap_do = cap_do   

    def ThoiGianXuatBan(self):
        return "Xuat ban thang 3,4,5/2026" 
    
    def __str__(self):
        return f"{super.__str__()}, Chuyen nganh: {self.chuyen_nganh}, Cap do: {self.cap_do}"
```

```python
class StackBook():
    def __init__(self):
        self.dsStack = []
    
    def push(self, b):
        self.dsStack.append(b)

    def isEmpty(self):
        return len(self.dsStack) == 0
    
    def pop(self):
        if self.isEmpty() :
            return None
        return self.dsStack.pop()
    
    def peek(self):
        if self.isEmpty():
            return None
        return self.dsStack[-1]
    
```

```python
class QueueBook:
    def __init__(self):
        self.dsQueue = []

    def is_empty(self):
        return len(self.dsQueue) == 0

    def enqueue(self, tai_lieu):
        self.dsQueue.append(tai_lieu)

    def dequeue(self):
        if self.is_empty():
            return None
        return self.dsQueue.pop(0)

    def front(self):
        if self.is_empty():
            return None
        return self.dsQueue[0]
```

```python
import json


class BookManagement:
    def __init__(self):
        self.dsBook = []

    def AddBook(self, book):
        self.dsBook.append(book)
    
    def Add_Book(self, soluong):
        for i in range(soluong):
            loai = int(input("Nhap loai: (1.Sach, 2.Sach Giao Khoa, 3.Sach Tham Khao)"))
            ma = input("Nhap ma:")  
            hoten = input("Nhap ten:")
            tac_gia = input("Nhap ten tac gia:")
            gia = float(input("Nhap gia:"))
            nam_xb = int(input("Nhap nam xb:"))

            if loai == 2:
                lop = input("Nhap lop:")
                mon = input("Nhap mon:")
                self.dsBook.append(SachGiaoKhoa(ma, hoten, tac_gia, gia, nam_xb, lop, mon))

            elif loai == 3:
                chuyen_nganh = input("Nhap chuyen nganh:")
                cap_do = input("Nhap cap do:")
                self.dsBook.append(SachThamKhao(ma, hoten, tac_gia, gia, nam_xb, chuyen_nganh, cap_do))

            else:
                self.dsBook.append(Book(ma, hoten, tac_gia, gia, nam_xb))

    def DelBook(self, ma):
        for b in self.dsBook:
            if b.ma == ma:
                self.dsBook.remove(b)
                return True  
        return False
    
    def DisplayBook(self):
        for b in self.dsBook:
            print(b)

    def SearchBook(self, ma):
        for b in self.dsBook:
            if b.ma == ma:
                return b
        return None
    
    def UpdateBook(self, ma, hoten, tac_gia, gia, nam_xb):
        for b in self.dsBook:
            if b.ma == ma:
                b.hoten = hoten
                b.tac_gia = tac_gia
                b.gia = float(gia)
                b.nam_xb = int(nam_xb)
                return True
        return False
    
    def SearchGia(self, gia):
        kq = [s for s in self.dsBook if s.gia == gia]
        return kq

    def SearchKhoangGia(self, min_gia, max_gia):
        kq = [s for s in self.dsBook if min_gia <= s.gia <= max_gia]
        return kq

    def tim_theo_nam_xb(self, nam):
        return [s for s in self.dsBook if s.nam_xb == nam]

    def tim_theo_tu_khoa_ten(self, tu_khoa):
        tu_khoa = tu_khoa.lower()
        return [s for s in self.dsBook if tu_khoa in s.ten.lower()]

    def BubbleSort(self):
        arr = self.dsBook
        for i in range(len(arr)):
            for j in range(0, len(arr)-1-i):
                if arr[j].gia > arr[j+1].gia:
                    arr[j], arr[j+1] = arr[j+1], arr[j]

    def SelectionSort(self):
        arr = self.dsBook
        n = len(arr)
        for i in range(n):
            min_index = i
            for j in range(i+1, n):
                if arr[min_index].ma > arr[j].ma:
                    min_index = j
            arr[i], arr[min_index] = arr[min_index], arr[i]

    def InsertionSort(self):
        arr = self.dsBook
        n = len(arr)
        for i in range (1, n):
            j = i-1
            while j >= 0 and arr[i].ten < arr[j].ten:
                arr[j+1] = arr[j]
                j -=1
            arr[j+1] = arr[i]

    def merge_sort(self):
        arr = self.dsBook
        if len(arr) <= 1:
            return arr
        
        mid = len(arr)//2
        left = arr[:mid]
        right = arr[mid:]

        left_sort = merge_sort(left)
        right_sort = merge_sort(right)

        merge_sort = merge(left_sort, right_sort)
        return merge_sort
    
    def merge(left_sort, right_sort):
        sorted = []
        i = j = 0

        while i < len(left_sort) and j < len(right_sort):
            if left_sort[i].nam_xb < right_sort[j].nam_xb:
                sorted.append(left_sort[i])
                i += 1
            else:
                sorted.append(right_sort[j])
                j += 1
        
        while i < len(left_sort):
            sorted.append(left_sort[i])
            i += 1
        while j < len(right_sort):
            sorted.append(right_sort[j])
            j += 1

    def QuickSort(arr):
        if len(arr) <= 1:
            return arr
        
        pivot = arr[len(arr)//2].gia

        left  = [x for x in arr if x.gia < pivot]
        mid   = [x for x in arr if x.gia == pivot]
        right = [x for x in arr if x.gia > pivot]

        return QuickSort(left) + mid + QuickSort(right)
    

    def save_to_json(self, filename="books.json"):
        data = []
        for b in self.dsBook:
            data.append({
                "ma": b.ma,
                "ten": b.ten,
                "tac_gia": b.tac_gia,
                "gia": b.gia,
                "nam_xb": b.nam_xb
            })
        
        with open(filename, "w", encoding="utf-8") as f:
            json.dump(data, f, ensure_ascii=False, indent=4)
        
        print("Da luu file JSON thanh cong")



    def load_from_json(self, filename="books.json"):
        self.dsBook.clear()
        
        with open(filename, "r", encoding="utf-8") as f:
            data = json.load(f)
        
        for item in data:
            b = Book(
                item["ma"],
                item["ten"],
                item["tac_gia"],
                float(item["gia"]),
                int(item["nam_xb"])
            )
            self.dsBook.append(b)
        
        print("Da doc file JSON thanh cong")
```

```python
def menu():
    ql = BookManagement()

    while True:
        print("\n========= BOOK MANAGEMENT =========")
        print("1. Thêm sách")
        print("2. Xóa sách theo mã")
        print("3. Hiển thị danh sách sách")
        print("4. Tìm sách theo mã")
        print("5. Cập nhật thông tin sách")
        print("6. Tìm kiếm linh hoạt (giá / năm / tên)")
        print("7. Sắp xếp Bubble theo giá")
        print("8. Sắp xếp Selection theo mã")
        print("9. Sắp xếp Insertion theo tên")
        print("10. Sắp xếp Merge theo năm xuất bản")
        print("11. Sắp xếp Quick theo giá")
        print("12. Đọc dữ liệu từ file JSON")
        print("13. Lưu dữ liệu ra file JSON")
        print("14. Demo StackSach")
        print("15. Demo QueueSach")
        print("0. Thoát")

        try:
            choice = int(input("Chọn chức năng: "))
        except:
            print("Vui lòng nhập số")
            continue

        if choice == 1:
            try:
                n = int(input("Nhập số lượng sách cần thêm: "))
                ql.Add_Book(n)
                print("Thêm sách thành công")
            except:
                print("Lỗi nhập dữ liệu")

        elif choice == 2:
            ma = input("Nhập mã sách cần xóa: ")
            if ql.DelBook(ma):
                print("Đã xóa sách")
            else:
                print("Không tìm thấy sách")

        elif choice == 3:
            ql.DisplayBook()

        elif choice == 4:
            ma = input("Nhập mã sách: ")
            b = ql.SearchBook(ma)
            if b:
                print(b)
            else:
                print("Không tìm thấy")

        elif choice == 5:
            ma = input("Nhập mã sách cần cập nhật: ")
            ten = input("Tên mới: ")
            tac_gia = input("Tác giả mới: ")
            gia = float(input("Giá mới: "))
            nam = int(input("Năm XB mới: "))
            if ql.UpdateBook(ma, ten, tac_gia, gia, nam):
                print("Cập nhật thành công")
            else:
                print("Không tìm thấy sách")

        elif choice == 6:
            print("1. Tìm theo giá chính xác")
            print("2. Tìm theo khoảng giá")
            print("3. Tìm theo năm XB")
            print("4. Tìm theo từ khóa tên")
            sub = input("Chọn: ")

            if sub == "1":
                gia = float(input("Nhập giá: "))
                kq = ql.SearchGia(gia)
            elif sub == "2":
                g1 = float(input("Giá min: "))
                g2 = float(input("Giá max: "))
                kq = ql.SearchKhoangGia(g1, g2)
            elif sub == "3":
                nam = int(input("Năm XB: "))
                kq = ql.tim_theo_nam_xb(nam)
            elif sub == "4":
                kw = input("Từ khóa tên: ")
                kq = ql.tim_theo_tu_khoa_ten(kw)
            else:
                print("Lựa chọn sai")
                continue

            if kq:
                for b in kq:
                    print(b)
            else:
                print("Không có kết quả")

        elif choice == 7:
            ql.BubbleSort()
            print("Đã sắp xếp Bubble theo giá")

        elif choice == 8:
            ql.SelectionSort()
            print("Đã sắp xếp Selection theo mã")

        elif choice == 9:
            ql.InsertionSort()
            print("Đã sắp xếp Insertion theo tên")

        elif choice == 10:
            ql.merge_sort()
            print("Đã sắp xếp Merge theo năm XB")

        elif choice == 11:
            ql.QuickSort()
            print("Đã sắp xếp Quick theo giá")

        elif choice == 12:
            ql.load_from_json()
            print("Đã đọc file")

        elif choice == 13:
            ql.save_to_json()
            print("Đã lưu file")

        elif choice == 14:
            print("Demo StackSach")
            stack = []
            for b in ql.dsBook:
                stack.append(b)
            print("Stack hiện tại:")
            for b in reversed(stack):
                print(b)

        elif choice == 15:
            print("Demo QueueSach")
            queue = []
            for b in ql.dsBook:
                queue.append(b)
            print("Queue hiện tại:")
            for b in queue:
                print(b)

        elif choice == 0:
            print("Thoát chương trình")
            break

        else:
            print("Lựa chọn không hợp lệ")
```
