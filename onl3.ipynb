```python
class SinhVien:
    def __init__(self, masv, hoten, diembigdata = None, diemctdl = None):
        self.masv = masv
        self.hoten = hoten
        self.diembigdata = diembigdata
        self.diemctdl = diemctdl

    def __str__(self):
        return f"{self.masv} - {self.hoten} - {self.diembigdata} - {self.diemctdl}"
    
    def diemtb(self):
        return (self.diembigdata + self.diemctdl) / 2
```

```python
class SVKSTN(SinhVien):
    def __init__(self, masv, hoten, diembigdata, diemctdl, diemav3):
        super().__init__(masv, hoten, diembigdata, diemctdl)
        self.diemav3 = diemav3

    def __str__(self):
        return super().__str__() + f" - {self.diemav3}"
    
    def diemtb(self):
        return super().diemtb() + self.diemav3
```

```python
class SVCLC(SinhVien):
    def __init__(self, masv, hoten, diembigdata, diemctdl, diemav4):
        super().__init__(masv, hoten, diembigdata, diemctdl)
        self.diemav4 = diemav4

    def __str__(self):
        return super().__str__() + f" - {self.diemav4}"

    def diemtb(self):
        return super().diemtb() + self.diemav4
```

```python
lst = []

def themSV():
    soluong = int(input("Nhap so luong sinh vien:"))
    for i in range(soluong):
        loai = int(input("Nhap loai(1. Sinh vien, 2. SVKSTN, 3. SVCLC):"))
        masv = input("Nhap masv:")
        hoten = input("Nhap hoten:")
        diembigdata = input("Nhap diem big data:")
        diemctdl = input("Nhap diem ctdl:")

        if loai == 2:
            diemav3 = input("Nhap diem av3:")
            lst.append(SVKSTN(masv, hoten, diembigdata, diemctdl, diemav3))

        elif loai == 3:
            diemav4 = input("Nhap diem av4:")
            lst.append(SVCLC(masv, hoten, diembigdata, diemctdl, diemav4))

        else:
            lst.append(SinhVien(masv, hoten, diembigdata, diemctdl))

def hienthi():
    for i in lst:
        print(i)
```

```python
def BubbleSort(arr):
    n = len(arr)

    for i in range(n):
        for j in range(0, n - 1 - i):
            if arr[j].masv > arr[j + 1].masv:
                arr[j], arr[j+1] = arr[j+1], arr[j]
```

```python
themSV()

hienthi()
```

```python
BubbleSort(lst)
hienthi()
```

```python
def WriteFile(path):
    with open(path, "w", encoding= "utf-8") as file:
        for sv in lst:
            file.write(str(sv) + "\n")
```
