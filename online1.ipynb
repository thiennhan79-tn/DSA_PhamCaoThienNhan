** tuyến tính: độ phức tạp O(n^2)

- Bubble sort: sắp xếp nổi bọt

Nguyên lí hoạt động: đẩy phần tử lớn về cuối cùng


```python
arr = [1, 15, 30, 58, 9]
```

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n-1-i):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
```

- selection sort

Nguyên lí hoạt động: tìm thằng nhỏ về đầu mảng

```python
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(i+1, n):
            if arr[i] > arr[j]:
                arr[i], arr[j] = arr[j], arr[j]
```

```python
def selection_sort1(arr):
    n = len(arr)
    for i in range(n):
        min_index = i
        for j in range(i+1, n):
            if arr[min_index] > arr[j]:
                min_index = j
        arr[i], arr[min_index] = arr[min_index], arr[i]
    return arr
```

```python
print(selection_sort1(arr))
```

- insertion sort

Nguyên lí hoạt động: chèn giá trị đúng vị trí

```python
def insertion_sort(arr):
    n = len(arr)
    for i in range (1, n):
        j = i-1
        while j >= 0 and arr[i] < arr[j]:
            arr[j+1] = arr[j]
            j -=1
        arr[j+1] = arr[i]
```

```python
def merge_sort(arr):
    #1. Tim dieu kien goc
    if len(arr) <= 1:
        return arr
    
    #2. Tim index, left, right
    mid = len(arr)//2
    left = arr[:mid]
    right = arr[mid:]

    #3. de quy: sap xep lan luot ben trai
    left_sort = merge_sort(left)
    right_sort = merge_sort(right)

    merge_sort = merge(left_sort, right_sort)

    return merge_sort
```

```python
def merge(left_sort, right_sort):
    sorted = []
    i = j = 0

    while i < len(left_sort) and j < len(right_sort):
        if left_sort[i] < right_sort[j]:
            sorted.append(left_sort[i])
            i += 1
        else:
            sorted.append(right_sort[j])
            j += 1
    
    #kiem tra cac gia tri con sot
    while i < len(left_sort):
        sorted.append(left_sort[i])
        i += 1
    while j < len(right_sort):
        sorted.append(right_sort[j])
        j += 1
```
