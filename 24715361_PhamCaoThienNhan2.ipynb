```python
import math
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None
    
    def add_node(self, value):
        if value < self.data:
            if self.left is None:
                self.left = Node(value)
            else:
                self.left.add_node(value)
        elif value > self.data:
            if self.right is None:
                self.right = Node(value)
            else:
                self.right.add_node(value)
    
    def NLR(self):
        print(self.data, end = " ")
        if self.left:
            self.left.NLR()
        if self.right:
            self.right.NLR()

    def NRL(self):
        print(self.data, end = " ")
        if self.right:
            self.right.NRL()
        if self.left:
            self.left.NRL()

    def LNR(self):
        if self.left:
            self.left.LNR()
        print(self.data, end = " ")
        if self.right:
            self.right.LNR()

    def RNL(self):
        if self.right:
            self.right.RNL()
        print(self.data, end = " ")
        if self.left:
            self.left.RNL()
    
    def LRN(self):
        if self.left:
            self.left.LRN()
        if self.right:
            self.right.LRN()
        print(self.data, end = " ")
    
    def RLN(self):
        if self.right:
            self.right.RLN()
        if self.left:
            self.left.RLN()
        print(self.data, end = " ")

    def is_prime(self, n):
        if n < 2:
            return False
        for i in range(2, int(math.sqrt(n)) + 1):
            if n % i == 0:
                return False
        return True

    def count_prime(self):
        count = 1 if self.is_prime(self.data) else 0
        
        if self.left:
            count += self.left.count_prime()
        if self.right:
            count += self.right.count_prime()
        
        return count

    def is_square(self, n):
        if n < 0:
            return False
        can = int(math.sqrt(n))
        return can * can == n

    def count_square(self):
        count = 1 if self.is_square(self.data) else 0
        
        if self.left:
            count += self.left.count_square()
        if self.right:
            count += self.right.count_square()
        
        return count
    
    def count_chan(self):
        count = 0
        
        if self.data % 2 == 0:
            count += 1
        
        if self.left:
            count += self.left.count_chan()
        
        if self.right:
            count += self.right.count_chan()
        
        return count

    def count_le(self):
        count = 0
        
        if self.data % 2 != 0:
            count += 1
        
        if self.left:
            count += self.left.count_le()
        
        if self.right:
            count += self.right.count_le()
        
        return count
    
    def find_max(self):
        if self.right:
            return self.right.find_max()
        return self
    
    def find_min(self):
        if self.left:
            return self.left.find_min()
        return self

    def node_2_con(self):
        if self.left and self.right:
            print(self.data, end = " ")
        if self.left:
            self.left.node_2_con()
        if self.right:
            self.right.node_2_con()
    
    def node_1_con(self):
        if (self.left is None) != (self.right is None):
            print(self.data, end = " ")
        if self.left:
            self.left.node_1_con()
        if self.right:
            self.right.node_1_con()

    def node_la(self):
        if self.left is None and self.right is None:
            print(self.data, end = " ")
        if self.left:
            self.left.node_la()
        if self.right:
            self.right.node_la()
    
    def search(self, value):
        if value == self.data:
            return True
        if self.left:
            self.left.search(value)
        if self.right:
            self.right.search(value)
        return False

    def delete(self, value):
        if value < self.data:
            if self.left:
                self.left = self.left.delete(value)
        
        elif value > self.data:
            if self.right:
                self.right = self.right.delete(value)
        
        else:
            if self.left is None and self.right is None:
                return None
            
            if self.left is None:
                return self.right
            if self.right is None:
                return self.left
            
            min_node = self.right.find_min()
            self.data = min_node.data
            self.right = self.right.delete(min_node.data)
        
        return self


    def read_from_file(filename):
        with open(filename, "r") as f:
            data = f.read().split()
        numbers = [int(x) for x in data] 
        if not numbers:
            return None
        root = Node(numbers[0])
        for n in numbers[1:]:
            root.add_node(n)
        return root
    
    def write_inorder_to_file(root, filename):
        def inorder(node, result):
            if node:
                inorder(node.left, result)
                result.append(str(node.data))
                inorder(node.right, result)
        
        result = []
        inorder(root, result)
        
        with open(filename, "w") as f:
            f.write(" ".join(result))
```

```python
root = Node.read_from_file("BinaryTreeSearch.txt")

print("NLR:")
root.NLR()

print("\nNRL:")
root.NRL()

print("\nLNR:")
root.LNR()

print("\nRNL:")
root.RNL()

print("\nLRN:")
root.LRN()

print("\nRLN:")
root.RLN()

print("\nSo nguyen to:", root.count_prime())
print("\nSo chinh phuong:", root.count_square())
print("\nSo chan:", root.count_chan())
print("\nSo le:", root.count_le())

print("\nMax:", root.find_max().data)
print("\nMin:", root.find_min().data)

print("\nNode 2 con:")
root.node_2_con()

print("\nNode 1 con:")
root.node_1_con()

print("\nNode la:")
root.node_la()

print("\nTim 12:", root.search(12))

root = root.delete(12)

Node.write_inorder_to_file(root, "output.txt")
```
