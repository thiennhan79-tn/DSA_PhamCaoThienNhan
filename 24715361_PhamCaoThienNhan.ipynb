```python
class MonHoc():
    def __init__(self, maMH, tenMH, soTC, soluong):
        self.maMH = maMH
        self.tenMH = tenMH
        self.soTC = soTC
        self.soluong = int(soluong)
        self.next = None
```

```python
class Stack:
    def __init__(self):
        self.lstStack = []
    
    def is_empty(self):
        return len(self.lstStack) == 0
    
    def push(self, monhoc):
        self.lstStack.append(monhoc)

    def pop(self):
        return self.lstStack.pop()
```

```python
class LinkedList():
    def __init__(self):
        self.head = None
    
    def is_empty(self):
        return self.head is None
    
    def print(self):
        cur = self.head
        while cur:
            print(f"{cur.maMH} - {cur.tenMH} - {cur.soTC} - {cur.soluong}", end = "\n")
            cur = cur.next
    
    def trungMH(self, maMH):
        cur = self.head
        while cur:
            if cur.maMH == maMH:
                return True
            cur = cur.next
        return False

    def add_first(self, maMH, tenMH, soTC, soluong):
        new_node = MonHoc(maMH, tenMH, soTC, soluong)
        new_node.next = self.head
        self.head = new_node

    def add_last(self, maMH, tenMH, soTC, soluong):
        new_node = MonHoc(maMH, tenMH, soTC, soluong)
        if self.is_empty():
            self.head = new_node
            return
        cur = self.head
        while cur.next:
            cur = cur.next
        cur.next = new_node

    def add_after_node(self, maMH_before, maMH, tenMH, soTC, soluong):
        cur = self.head
        while cur:
            if cur.maMH == maMH_before:
                new_node = MonHoc(maMH, tenMH, soTC, soluong)
                new_node.next = cur.next
                cur.next = new_node
                return
            cur = cur.next
    
    def find(self, maMH):
        cur = self.head
        while cur:
            if cur.maMH == maMH:
                return cur
            cur = cur.next

        return None
    
    def update(self, maMH, tenMH = None, soTC = None, soluong = None):
        node = self.find(maMH)
        if node is None:
            return False
        if tenMH is not None:
            node.tenMH = tenMH
        if soTC is not None:
            node.soTC = soTC
        if soluong is not None:
            node.soluong = soluong
        return True
            
        
    def delete(self, maMH):
        if self.is_empty():
            return False
        if self.head.maMH == maMH:
            self.head = self.head.next
            return True
        cur  = self.head
        while cur.next:
            if cur.next.maMH == maMH:
                cur.next = cur.next.next
                return True
            cur = cur.next
        return False
    
    def read_file(self, path):
        with open(path, 'r', encoding='utf-8') as f:
            for line in f:
                row = [ row.strip() for row in line.split(', ') ]
                maMH, tenMH, soTC, soluong = row[0], row[1], row[2], row[3]
                self.add_last(maMH, tenMH, soTC, soluong)

    def find_SLLH30(self):
        stk = Stack()
        cur = self.head
        while cur:
            if int(cur.soluong) > 30:
                stk.push(cur)
            cur = cur.next
        return stk
    
    def write_SLLH30_to_file(self, path):
        stk = self.find_SLLH30()
        with open(path, 'w', encoding='utf-8') as f:
            while not stk.is_empty():
                mh = stk.pop()
                f.write(f"{mh.maMH} - {mh.tenMH} - {mh.soTC} - {mh.soluong}\n")
    
    def write_all_to_file(self, path):
        with open(path, "w", encoding="utf-8") as f:
            cur = self.head
            while cur:
                f.write(f"{cur.maMH} - {cur.tenMH} - {cur.soTC} - {cur.soluong}\n")
                cur = cur.next
```

```python
def menu():
    lst = LinkedList()

    while True:
        print("\n===== MENU =====")
        print("1. Doc file")
        print("2. Hien thi danh sach")
        print("3. Them vao dau")
        print("4. Them vao cuoi")
        print("5. Them sau maMH")
        print("6. Tim kiem theo maMH")
        print("7. Xoa theo maMH")
        print("8. Tim mon hoc co so luong > 30")
        print("9. In ket qua mon hoc > 30 ra file output.txt")
        print("10. Cap nhat")
        print("11. Xuat danh sach")
        print("0. Thoat")

        choice = input("Chon chuc nang: ")

        if choice == '1':
            path = input("Nhap duong dan file: ")
            lst.read_file(path)
            print("Da doc file!")

        elif choice == '2':
            lst.print()

        elif choice == '3':
            ma = input("Ma MH: ")
            ten = input("Ten MH: ")
            tc = input("So TC: ")
            sl = input("So luong: ")
            lst.add_first(ma, ten, tc, sl)

        elif choice == '4':
            ma = input("Ma MH: ")
            ten = input("Ten MH: ")
            tc = input("So TC: ")
            sl = input("So luong: ")
            lst.add_last(ma, ten, tc, sl)

        elif choice == '5':
            ma_before = input("Nhap maMH truoc: ")
            ma = input("Ma MH moi: ")
            ten = input("Ten MH: ")
            tc = input("So TC: ")
            sl = input("So luong: ")
            lst.add_after_node(ma_before, ma, ten, tc, sl)

        elif choice == '6':
            ma = input("Nhap maMH can tim: ")
            mh = lst.find(ma)
            if mh:
                print(f"{mh.maMH} - {mh.tenMH} - {mh.soTC} - {mh.soluong}")
            else:
                print("Khong tim thay!")

        elif choice == '7':
            ma = input("Nhap maMH can xoa: ")
            if lst.delete(ma):
                print("Da xoa!")
            else:
                print("Khong tim thay!")

        elif choice == '8':
            stk = lst.find_SLLH30()
            if stk.is_empty():
                print("Khong co mon nao > 30")
            else:
                print("Danh sach mon > 30:")
                while not stk.is_empty():
                    mh = stk.pop()
                    print(f"{mh.maMH} - {mh.tenMH} - {mh.soTC} - {mh.soluong}")

        elif choice == '9':
            lst.write_SLLH30_to_file("output.txt")
            print("Da ghi file output.txt")
        
        elif choice == '10':
            ma = input("Nhap maMH can cap nhat: ")
            ten = input("Ten moi (bo trong neu khong doi): ")
            tc = input("So TC moi (bo trong neu khong doi): ")
            sl = input("So luong moi (bo trong neu khong doi): ")

            ten = ten if ten != "" else None
            tc = tc if tc != "" else None
            sl = sl if sl != "" else None

            if lst.update(ma, ten, tc, sl):
                print("Cap nhat thanh cong!")
            else:
                print("Khong tim thay!")

        elif choice == '11':
            lst.write_all_to_file("f.txt")
            print("Da ghi file f.txt")
                
        elif choice == '0':
            print("Thoat chuong trinh!")
            break

        else:
            print("Lua chon khong hop le!")
```

OOP gồm kế thừa, đa hình, trừu tượng, đóng gói.

Stack hoạt động theo LIFO, Queue theo FIFO.

Linked List là cấu trúc dữ liệu dạng node liên kết.

Có 3 loại DSLK: đơn, kép, vòng.

Các thuật toán sắp xếp gồm Bubble, Selection, Insertion (O(n²)) và Merge, Quick (O(n log n)).

Stack = chồng sách → lấy trên cùng

Queue = xếp hàng → lấy người đầu

Override là ghi đè phương thức của lớp cha trong lớp con.

Overloading là nạp chồng hàm cùng tên nhưng khác tham số.
