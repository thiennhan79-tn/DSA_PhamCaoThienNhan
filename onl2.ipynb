```python
class Node:
    def __init__(self, masv, hoten, gpa):
        self.masv = masv 
        self.hoten = hoten
        self.gpa = gpa
        self.next = None
```

```python
class LinkedList:
    def __init__(self):
        self.head = None

    def is_empty(self):
        return self.head is None
    
    def print(self):
        if self.is_empty():
            print("Danh sach lien ket rong")
            return 
        cur = self.head
        while cur:
            print(f"{cur.masv} - {cur.hoten} - {cur.gpa}", end=" \n ")
            cur = cur.next
        print()

    # day vao dau dsach
    def add_first(self, masv, hoten, gpa):
        new_node = Node(masv, hoten, gpa)
        new_node.next = self.head
        self.head = new_node

    # day vao cuoi dsach
    def add_last(self, masv, hoten, gpa):
        new_node = Node(masv, hoten, gpa)
        #th1:dsach rong:
        if self.is_empty():
            self.head = new_node
            return 
        #th2:dsach khac rong
        cur = self.head
        while cur.next:
            cur = cur.next
        cur.next = new_node
        
    #day vao vi tri bat ki
    def add_after_node(self, masv_before, masv, hoten, gpa):
        cur = self.head
        while cur:
            if cur.masv == masv_before:
                new_node = Node(masv, hoten, gpa)
                new_node.next = cur.next
                cur.next = new_node
                return
            cur = cur.next

    def find_number(self, masv):
        cur = self.head
        while cur:
            if cur.masv == masv:
                #return cur
                return True
            cur = cur.next
            
        return False
    
    def find(self, data):
        cur = self.head
        while cur:
            if cur.data == data:
                return cur
            cur = cur.next
            
        return None
    
    def find_node_chan(self):
        lst = []
        cur = self.head
        while cur:
            if cur.data % 2 == 0:
               lst.append(cur.data)
            cur = cur.next
        return lst
    
    def update(self, data_old, data_new):
        node = self.find(data_old)
        if node is None:
            return False
        
        node.data = data_new
        return True
    
    def delete(self, data):
        if self.is_empty():
            return False
        
        #th1: nam o dau dsach
        if self.head.data == data:
            self.head = self.head.next
            return True
        
        #th2: nam o giua dsach hoac cuoi
        cur  = self.head
        while cur.next:
            if cur.data == data:
                cur.next = cur.next.next
                return True
            cur = cur.next

        #th3: ra khoi dsach van khong tim thay
        return False

    def save_file(self, filename):
        if self.is_empty():
            return
        with open(filename, "w") as file:
            cur = self.head
            while cur:
                file.write(f"{cur.data}\n")
                cur = cur.next

    def read_file(self, filename):
        with open(filename, "r") as file:
            for line in file:
                line = line.strip()
                for number in line.split(","):
                    self.add_first(number)
```

```python
lst = LinkedList()
lst.add_last("SV01", "Nhan", 10)
lst.add_first("SV02", "Hung", 9)
lst.print()
```
