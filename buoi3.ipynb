```python
from abc import ABC, abstractmethod

class SinhVien(ABC):
    uni = "Truong dh Cong Nghiep tphcm"
    def __init__(self, mssv, hoten):
        self.mssv = mssv
        self.hoten = hoten
        self.__cccd = '123'

    def get_cccd(self):
        return self.__cccd

    def set_cccd(self, cccd):
        self.__cccd = cccd
    
    def InThongTin(self):
        print(f"{self.mssv} - {self.hoten} - {self.uni}")
    
    @abstractmethod
    def TinhDTB(self):
        pass
```

```python
sv1 = SinhVien('12345', 'Thien Nhan')
sv2 = SinhVien('12346', 'Hung')
```

```python
sv1.InThongTin()
sv2.InThongTin()
```

```python
class KSTN(SinhVien):
    def __init__(self, mssv, hoten, diemCTRR, diemTA):
        super().__init__(mssv, hoten)
        self.diemCTRR = diemCTRR
        self.diemTA = diemTA
    
    def InThongTin(self):
        super().InThongTin()
        print(f"Diem CTRR :{self.diemCTRR} - Diem TA :{self.diemTA}")
    
    def TinhDTB(self):
        dtb = (self.diemCTRR + self.diemTA)/2
        print(f"Diem tb: {dtb}")
```

```python
ks1 = KSTN('SV1', 'Thien Nhan', 8.0, 9.0)
ks1.InThongTin()
ks1.TinhDTB()

ks1.get_cccd()
```

```python
class TCTA(SinhVien):
    def __init__(self, mssv, hoten, diemAV3, diemAV4):
        super().__init__(mssv, hoten)
        self.diemAV3 = diemAV3
        self.diemAV4 = diemAV4
    
    def InThongTin(self):
        super().InThongTin()
        print(f"Diem AV3: {self.diemAV3} - DiemAV4: {self.diemAV4}")

    def TinhDTB(self):
        dtb = (self.diemAV3 + self.diemAV4)/2
        print(f"Diem trung binh: {dtb}")
```

```python
tcta1 = TCTA('1234', 'Thien Nhan', 8, 9)
tcta1.InThongTin()
tcta1.TinhDTB()
```
