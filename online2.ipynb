```python
arr = [15, 3, 6, 2, 1]
len(arr)
```

```python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    
    #2. index pivot
    pivot = arr[len(arr)//2]

    #3. < pivot, > pivot, = pivot
    left = [x for x in arr if x < pivot] #list comprehension
    mid = [x for x in arr if x == pivot]
    right = [x for x in arr if x > arr]

    #4. gop
    return quick_sort(left) + mid + quick_sort(right)

arr = [15, 3, 6, 2, 1]
arr_new = quick_sort(arr)
arr_new

```
