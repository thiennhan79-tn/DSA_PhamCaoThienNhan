```python
class Node:
    def __init__(self, data):
        self.data = data 
        self.next = None
```

```python
class LinkedList:
    def __init__(self):
        self.head = None

    def is_empty(self):
        return self.head is None
    
    def print(self):
        if self.is_empty():
            print("Danh sach lien ket rong")
            return 
        cur = self.head
        while cur:
            print(cur.data, end = " ")
            cur = cur.next
        print()

    # day vao dau dsach
    def add_first(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    # day vao cuoi dsach
    def add_last(self, data):
        new_node = Node(data)
        #th1:dsach rong:
        if self.is_empty():
            self.head = new_node
            return 
        #th2:dsach khac rong
        cur = self.head
        while cur.next:
            cur = cur.next
        cur.next = new_node
        
    #day vao vi tri bat ki
    def add_after_node(self, data_before, data):
        cur = self.head
        while cur:
            if cur.data == data_before:
                new_node = Node(data)
                new_node.next = cur.next
                cur.next = new_node
                return
            cur = cur.next

    def find_number(self, data):
        cur = self.head
        while cur:
            if cur.data == data:
                #return cur
                return True
            cur = cur.next
            
        return False
    
    def find(self, data):
        cur = self.head
        while cur:
            if cur.data == data:
                return cur
            cur = cur.next
            
        return None
    
    def find_node_chan(self):
        lst = []
        cur = self.head
        while cur:
            if cur.data % 2 == 0:
               lst.append(cur.data)
            cur = cur.next
        return lst
    
    def update(self, data_old, data_new):
        node = self.find(data_old)
        if node is None:
            return False
        
        node.data = data_new
        return True
    
    def delete(self, data):
        if self.is_empty():
            return False
        
        #th1: nam o dau dsach
        if self.head.data == data:
            self.head = self.head.next
            return True
        
        #th2: nam o giua dsach hoac cuoi
        cur  = self.head
        while cur.next:
            if cur.data == data:
                cur.next = cur.next.next
                return True
            cur = cur.next

        #th3: ra khoi dsach van khong tim thay
        return False

    def save_file(self, filename):
        if self.is_empty():
            return
        with open(filename, "w") as file:
            cur = self.head
            while cur:
                file.write(f"{cur.data}\n")
                cur = cur.next

    def read_file(self, filename):
        with open(filename, "r") as file:
            for line in file:
                line = line.strip()
                for number in line.split(","):
                    self.add_first(number)
```

```python
lst= LinkedList()

arr = [1,3,5,6]
for i in arr:
    lst.add_first(i)

lst.print()
print("--------")
lst.add_last(7)
lst.print()

print("-----")
lst.add_after_node(3, 99)
lst.print()


print("-----")
data = 33
kq = lst.find_number(data)
if kq == True:
    print(f"Tim thay {data} trong dslk")
else:
    print(f"Khong tim thay {data} trong dslk")

print("-----")
print("Node chan:")
lst_even = lst.find_node_chan()
for i in lst_even:
    print(i)

print("-----")
update_node = lst.update(6, 66)
if update_node == True:
    print("Cap nhat thanh cong")
else:
    print("Cap nhat that bai")

lst.print()

print("-----")
print("Xoa 3")
print(lst.delete(3))
print()


print("-----")
lst.save_file("nhan.txt")

lst.read_file("a.txt")

```
