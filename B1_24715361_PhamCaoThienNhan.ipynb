**Lưu ý:** Tên thuộc tính dùng **tiếng Việt không dấu**: `ma_tl`, `ten_tl`, `nam_xb`, `trang_thai`, `tac_gia`, ...

## 1) class

- **Class trừu tượng `TaiLieu`:** Lưu thông tin `ma_tl`, `ten_tl`, `nam_xb`, `trang_thai`.
  - Dùng **public attributes** (không bắt buộc private/getter/setter).
  - Validate dữ liệu trong `__init__`: `nam_xb >= 1900`, `trang_thai` thuộc `{San sang, Dang muon}`.

- **Kế thừa:** Ít nhất 2 lớp con từ `TaiLieu`:
  1. `Sach`: thêm `tac_gia`, `the_loai`.
  2. `TapChi`: thêm `so_phat_hanh`.

- Bắt buộc có `@abstractmethod` trong `TaiLieu`:
  - `tinh_phi_muon(so_ngay)`
  - `thong_tin()`

## 2) Stack/Queue + Abstract

- Tạo `QueueMuon` để quản lý hàng chờ mượn tài liệu (FIFO).
- Tạo `StackTra` để quản lý lịch sử trả tài liệu (LIFO).
- Dùng `abstractmethod` cho các hàm chính, ví dụ:
  - Queue: `enqueue`, `dequeue`, `front`, `is_empty`
  - Stack: `push`, `pop`, `peek`, `is_empty`

## 3) LibraryManagement Class

Quản lý danh sách tài liệu với các chức năng:
1. Thêm tài liệu (`Sach`/`TapChi`) bằng nhập tay hoặc từ file cho 1 loại.
2. Xóa tài liệu theo `ma_tl`.
3. Hiển thị danh sách tài liệu.
4. Tìm tài liệu theo `ma_tl`.
5. Cập nhật thông tin tài liệu.
6. Đăng ký mượn (`enqueue`) và xử lý mượn tiếp theo (`dequeue`).
7. Trả tài liệu và lưu vào stack lịch sử.
8. Xem lần trả gần nhất.

## 4) Sắp xếp (3 thuật toán)

Mỗi thuật toán một tiêu chí:
- **Bubble Sort** — theo `ten_tl` (A -> Z)
- **Selection Sort** — theo `nam_xb` (mới -> cũ)
- **Insertion Sort** — theo `ma_tl` (tăng dần)

> Không dùng `sorted()` để thay cho phần chính của thuật toán.

## 5) File I/O

- Đọc dữ liệu tài liệu từ file (CSV/JSON).
- Lưu danh sách hiện tại ra file (CSV/JSON).

## 6) Menu

Tạo hàm `menu()` chạy vòng lặp `while True`, cho người dùng chọn chức năng bằng số:
1. Thêm tài liệu
2. Xóa tài liệu theo mã
3. Hiển thị danh sách tài liệu
4. Tìm tài liệu theo mã
5. Cập nhật thông tin tài liệu
6. Đăng ký mượn (Queue)
7. Xử lý mượn tiếp theo (Queue)
8. Trả tài liệu (Stack)
9. Xem lần trả gần nhất (Stack)
10. Sắp xếp Bubble theo tên
11. Sắp xếp Selection theo năm xuất bản
12. Sắp xếp Insertion theo mã
13. Đọc dữ liệu từ file
14. Lưu dữ liệu ra file
0. Thoát

**Yêu cầu menu:**
- Có bắt lỗi nhập sai lựa chọn (`try/except`).
- Sau mỗi thao tác phải in kết quả hoặc thông báo thành công/thất bại.
- Chỉ thoát khi người dùng chọn `0`.

## 7) Yêu cầu chạy minh họa

- Tạo ít nhất 8 tài liệu, có cả `Sach` và `TapChi`.
- Mô phỏng ít nhất 4 lượt mượn và 2 lượt trả.
- In danh sách trước/sau cho cả 3 thuật toán sort.
- Có ít nhất 3 test lỗi nhập liệu (`nam_xb < 1900`, `trang_thai` sai, `so_ngay <= 0`).


```python
from abc import ABC, abstractmethod
class TaiLieu(ABC):
    def __init__(self, ma_tl, ten_tl, nam_xb, trang_thai):
        if nam_xb < 1900:
            raise ValueError("Nam xuat ban phai >= 1900")
        if trang_thai not in {"San sang", "Dang muon"}:
            raise ValueError("Trang thai phai la 'San sang' hoac 'Dang muon'")
        self.ma_tl = ma_tl
        self.ten_tl = ten_tl
        self.nam_xb = nam_xb
        self.trang_thai = trang_thai

    @abstractmethod
    def tinh_phi_muon(self, so_ngay):
        pass

    @abstractmethod
    def thong_tin(self):
        pass  
```

```python
class Sach(TaiLieu):
    def __init__(self, ma_tl, ten_tl, nam_xb, trang_thai, tac_gia, the_loai ):
        super().__init__(ma_tl, ten_tl, nam_xb, trang_thai)
        self.tac_gia = tac_gia
        self.the_loai = the_loai

    def tinh_phi_muon(self, so_ngay):
        if so_ngay <= 0:
            raise ValueError("So ngay muon phai > 0")
        return so_ngay * 10000
    
    def thong_tin(self):
        return f"Ma sach: {self.ma_tl}, Ten sach: {self.ten_tl}, Nam xb: {self.nam_xb}, Trang thai: {self.trang_thai}, Tac gia: {self.tac_gia}, The loai: {self.the_loai}"
```

```python
class TapChi(TaiLieu):
    def __init__(self, ma_tl, ten_tl, nam_xb, trang_thai, so_phat_hanh):
        super().__init__(ma_tl, ten_tl, nam_xb, trang_thai)
        self.so_phat_hanh = so_phat_hanh
        
    def tinh_phi_muon(self, so_ngay):
        if so_ngay <= 0:
            raise ValueError("So ngay muon phai > 0")
        return so_ngay * 5000
    
    def thong_tin(self):
        return f"Ma tap chi: {self.ma_tl}, Ten tap chi: {self.ten_tl}, Nam xb: {self.nam_xb}, Trang thai: {self.trang_thai}, So phat hanh: {self.so_phat_hanh}"
```

```python
class QueueMuon:
    def __init__(self):
        self.tlQueue = []

    def is_empty(self):
        return len(self.tlQueue) == 0

    def enqueue(self, tai_lieu):
        self.tlQueue.append(tai_lieu)

    def dequeue(self):
        if self.is_empty():
            return None
        return self.tlQueue.pop(0)

    def front(self):
        if self.is_empty():
            return None
        return self.tlQueue[0]
```

```python
class StackTra():
    def __init__(self):
        self.tlSack = []

    def is_empty(self):
        return len(self.tlSack) == 0

    def push(self, tai_lieu):
        self.tlSack.append(tai_lieu)

    def pop(self):
        if self.is_empty():
            return None
        return self.tlSack.pop()

    def peek(self):
        if self.is_empty():
            return None
        return self.tlSack[-1]
```

```python
import csv


class QuanLyTaiLieu:
    def __init__(self):
        self.ds_tai_lieu = []
        self.hang_cho_muon = QueueMuon()
        self.lich_su_tra = StackTra()

    def them_tai_lieu(self, soluong):
        for i in range(soluong):
            loai = int(input("Nhap loai: (1.Tai Lieu, 2.Sach, 3.Tap Chi)"))
            ma = input("Nhap ma:")
            ten = input("Nhap ten:")
            nam_xb = int(input("Nhap nam xb:"))
            trang_thai = input("Nhap trang thai:")

            if loai == 2:
                tac_gia = input("Nhap ten tac gia:")
                the_loai = input("Nhap the loai:")
                self.ds_tai_lieu.append(Sach(ma, ten, nam_xb, trang_thai, tac_gia, the_loai))

            if loai == 3:
                so_phat_hanh = input("Nhap so phat hanh:")
                self.ds_tai_lieu.append(TapChi(ma, ten, nam_xb, trang_thai, so_phat_hanh))

            else:
                self.ds_tai_lieu.append(TaiLieu(ma, ten, nam_xb, trang_thai))
    
    def xoa_tai_lieu(self, ma_tl):
        for tl in self.ds_tai_lieu:
            if tl.ma_tl == ma_tl:
                self.ds_tai_lieu.remove(tl)
                return "Da xoa tai lieu"
        return "Khong tim thay"

    def hien_thi(self):
        if not self.ds_tai_lieu:
            print("Danh sach rong")
            return
        for tl in self.ds_tai_lieu:
            print(tl.thong_tin())

    def tim_theo_ma(self, ma_tl):
        for tl in self.ds_tai_lieu:
            if tl.ma_tl == ma_tl:
                return tl
        return None

    def cap_nhat(self, ma_tl, ten=None, nam=None, trang_thai=None):
        tl = self.tim_theo_ma(ma_tl)
        if not tl:
            return "Khong tim thay"

        if ten:
            tl.ten_tl = ten
        if nam:
            tl.nam_xb = nam
        if trang_thai:
            tl.trang_thai = trang_thai

        return "Cap nhat thanh cong"
    
    def dang_ky_muon(self, ma_tl):
        tl = self.tim_theo_ma(ma_tl)
        if not tl:
            return "Khong tim thay tai lieu"

        self.hang_cho_muon.enqueue(tl)
        return "Da them vao hang cho muon"

    def xu_ly_muon(self):
        if self.hang_cho_muon.is_empty():
            return None
        return self.hang_cho_muon.dequeue()

    def tra_tai_lieu(self, ma_tl):
        tl = self.tim_theo_ma(ma_tl)
        if not tl:
            return "Khong tim thay tai lieu"

        self.lich_su_tra.push(tl)
        return "Da tra tai lieu"

    def xem_tra_gan_nhat(self):
        return self.lich_su_tra.peek()

    def bubble_sort_theo_ten(self):
        n = len(self.ds_tai_lieu)
        for i in range(n):
            for j in range(0, n - 1 - i):
                if self.ds_tai_lieu[j].ten_tl > self.ds_tai_lieu[j+1].ten_tl:
                    self.ds_tai_lieu[j], self.ds_tai_lieu[j+1] = self.ds_tai_lieu[j+1], self.ds_tai_lieu[j]

    def selection_sort_theo_nam(self):
        n = len(self.ds_tai_lieu)
        for i in range(n):
            max_idex = i
            for j in range(i+1, n):
                if self.ds_tai_lieu[j].nam_xb > self.ds_tai_lieu[max_idex].nam_xb:
                    max_idex = j
            self.ds_tai_lieu[i], self.ds_tai_lieu[max_idex] = self.ds_tai_lieu[max_idex], self.ds_tai_lieu[i]

    def insertion_sort_theo_ma(self):
        arr = self.ds_tai_lieu
        n = len(arr)
        for i in range (1, n):
            j = i-1
            while j >= 0 and arr[i].ma_tl < arr[j].ma_tl:
                arr[j+1] = arr[j]
                j -=1
            arr[j+1] = arr[i]


    def luu_csv(self, ten_file):
        with open(ten_file, "w", newline='', encoding="utf-8") as f:
            writer = csv.writer(f)
            writer.writerow(["loai", "ma", "ten", "nam_xb", "trang_thai", "extra1", "extra2"])

            for tl in self.ds_tai_lieu:
                if isinstance(tl, Sach):
                    writer.writerow(["Sach", tl.ma_tl, tl.ten_tl, tl.nam_xb, tl.trang_thai, tl.tac_gia, tl.the_loai])
                elif isinstance(tl, TapChi):
                    writer.writerow(["TapChi", tl.ma_tl, tl.ten_tl, tl.nam_xb, tl.trang_thai, tl.so_phat_hanh, ""])

    def doc_csv(self, ten_file):
        with open(ten_file, "r", encoding="utf-8") as f:
            reader = csv.reader(f)
            next(reader) 
            for row in reader:
                loai, ma, ten, nam, tt, e1, e2 = row
                nam = int(nam)
                if loai == "Sach":
                    tl = Sach(ma, ten, nam, tt, e1, e2)
                elif loai == "TapChi":
                    tl = TapChi(ma, ten, nam, tt, int(e1))
                else:
                    continue
                self.ds_tai_lieu.append(tl)



```

```python
def menu():
    ql = QuanLyTaiLieu()

    while True:
        print("\n===== MENU QUAN LY TAI LIEU =====")
        print("1. Them tai lieu")
        print("2. Xoa tai lieu theo ma")
        print("3. Hien thi danh sach tai lieu")
        print("4. Tim tai lieu theo ma")
        print("5. Cap nhat thong tin tai lieu")
        print("6. Dang ky muon (Queue)")
        print("7. Xu ly muon tiep theo (Queue)")
        print("8. Tra tai lieu (Stack)")
        print("9. Xem lan tra gan nhat (Stack)")
        print("10. Sap xep Bubble theo ten")
        print("11. Sap xep Selection theo nam xb")
        print("12. Sap xep Insertion theo ma")
        print("13. Doc du lieu tu file")
        print("14. Luu du lieu ra file")
        print("0. Thoat")

        try:
            chon = int(input("Nhap lua chon: "))
        except:
            print("Vui long nhap so hop le")
            continue

        if chon == 0:
            print("Tam biet!")
            break

        elif chon == 1:
            try:
                so = int(input("Nhap so luong tai lieu: "))
                ql.them_tai_lieu(so)
                print("Them tai lieu thanh cong")
            except:
                print("Loi khi them tai lieu")

        elif chon == 2:
            ma = input("Nhap ma tai lieu can xoa: ")
            print(ql.xoa_tai_lieu(ma))

        elif chon == 3:
            print("\n===== DANH SACH TAI LIEU =====")
            ql.hien_thi()

        elif chon == 4:
            ma = input("Nhap ma tai lieu can tim: ")
            tl = ql.tim_theo_ma(ma)
            if tl:
                print("Tim thay:")
                print(tl.thong_tin())
            else:
                print("Khong tim thay")

        elif chon == 5:
            ma = input("Nhap ma tai lieu can cap nhat: ")
            ten = input("Ten moi (bo trong neu khong doi): ")
            nam = input("Nam moi (bo trong neu khong doi): ")
            tt = input("Trang thai moi (bo trong neu khong doi): ")

            nam = int(nam) if nam else None
            ten = ten if ten else None
            tt = tt if tt else None

            print(ql.cap_nhat(ma, ten, nam, tt))

        elif chon == 6:
            ma = input("Nhap ma tai lieu muon: ")
            print(ql.dang_ky_muon(ma))

        elif chon == 7:
            tl = ql.xu_ly_muon()
            if tl:
                print("Dang xu ly muon:")
                print(tl.thong_tin())
            else:
                print("Hang cho rong")

        elif chon == 8:
            ma = input("Nhap ma tai lieu tra: ")
            print(ql.tra_tai_lieu(ma))

        elif chon == 9:
            tl = ql.xem_tra_gan_nhat()
            if tl:
                print("Lan tra gan nhat:")
                print(tl.thong_tin())
            else:
                print("Chua co lich su tra")

        elif chon == 10:
            ql.bubble_sort_theo_ten()
            print("Da sap xep theo ten (A-Z)")

        elif chon == 11:
            ql.selection_sort_theo_nam()
            print("Da sap xep theo nam xb (moi → cu)")

        elif chon == 12:
            ql.insertion_sort_theo_ma()
            print("Da sap xep theo ma TL (tang dan)")

        elif chon == 13:
            ten_file = input("Nhap ten file CSV: ")
            try:
                ql.doc_csv(ten_file)
                print("Doc file CSV thanh cong")
            except:
                print("Loi khi doc file CSV")

        elif chon == 14:
            ten_file = input("Nhap ten file CSV: ")
            try:
                ql.luu_csv(ten_file)
                print("Luu file CSV thanh cong")
            except:
                print("Loi khi luu file CSV")


        else:
            print("Lua chon khong hop le")
```
