```python
class KhoaHoc:
    def __init__(self, makh, tenkh, sotc):
        self.makh = makh
        self.tenkh = tenkh
        self.sotc = int(sotc)
        self.next = None

    def __str__(self):
        return f"{self.makh} - {self.tenkh} - {self.sotc}"
```

```python
class Stack():
    def __init__(self):
        self.dsStack = []

    def is_empty(self):
        return len(self.dsStack) == 0
    
    def push(self, kh):
        self.dsStack.append(kh)

    def pop(self):
        self.dsStack.pop()

    def printkh(self):
        for i in self.dsStack:
            print(i)
```

```python
class DSLK:
    def __init__(self):
        self.head = None

    def addKHvaodau(self, makh, tenkh, sotc):
        new_node = KhoaHoc(makh, tenkh, sotc)
        new_node.next = self.head
        self.head = new_node

    def hienthiKH(self):
        cur = self.head
        while cur:
            print(f"{cur.makh} - {cur.tenkh} - {cur.sotc}")
            cur = cur.next
    
    def importfile(self, path):
        with open(path, "r", encoding= "utf-8") as file:
            for line in file:
                line = line.strip()
                if not line:
                    continue

                parts = [row.strip() for row in line.split(', ')]
                makh, tenkh, sotc = parts[0], parts[1], parts[2]
                self.addKHvaodau(makh, tenkh, sotc)

    def findsotc4(self):
        self.stackkh = Stack()
        cur = self.head
        while cur:
            if cur.sotc >= 4:
                self.stackkh.push(cur)
            cur = cur.next

        return self.stackkh

    def printkh(self):
        self.stackkh.printkh()

    def find(self, name):
        cur = self.head
        while cur:
            if cur.tenkh == name:
                return cur
            cur = cur.next
        
        return None
```
