```python
from abc import ABC, abstractmethod
class Person(ABC):
    def __init__(self, id, name):
        self._id = id
        self._name = name

    def get_id(self):
        return self._id
    
    def get_name(self):
        return self._name
    
    def set_id(self, name):
        self._name = name

    @abstractmethod
    def get_role(self):
        pass

    @abstractmethod
    def to_dict(self):
        pass
```

```python
class Student(Person):
    def __init__(self, id, name, gpa):
        super().__init__(id, name)
        self.__gpa = gpa

    def set_gpa(self, gpa):
        self.__gpa = gpa
    
    def get_gpa(self):
        return self.__gpa
    
    def get_role(self):
        return "Student"
    
    def to_dict(self):
        return {
            "id": self._id(),
            "name": self._name(),
            "gpa": self.__gpa()
        }
```

```python
class UnderStudent(Student):
    def __init__(self, id, name, gpa, year):
        super().__init__(id, name, gpa)
        self.__year = year

    def get_role(self):
        return "UnderStudent"
    
    def to_dict(self):
        pass
```

```python
class GraduateStudent(Student):
    def __init__(self, id, name, gpa, thesis):
        super().__init__(id, name, gpa)
        self._thesis = thesis

    def get_role(self):
        return "GraduateStudent"
    
    def to_dict(self):
        pass
```

```python
class Manager:
    def __init__(self):
        self._lstStud = []

    def add_Student(self, s):
        self._lstStud.append(s)

    def list_Students(self):
        return self._lstStud
    
    def find_Student(self, id):
        for i in self._lstStud:
            if i.get_id() == id:
                return i
        return None
    
    def del_Student(self, id):
        s = self.find_Student(id)
        if s:
            del self._lstStud[s]
            return 

    def update_Student(self, id, name, gpa):
        s = self.find_Student(id)
        if s:
            if name:
                s.set_name(name) 
            if gpa:
                s.set_gpa(gpa)
            print(f"Da cap nhat sinh vien {id}")
        else:
            print(f"Khong tim thay sinh vien {id}")

    
    def print_Students(self):
        if not self._lstStud:
            print("Danh sách rỗng")
            return 
        print("Danh sach sinh viên:")
        print("----")
        for s in self._lstStud:
            print(f"Ma sinh vien: {s.get_id()}")
            print(f"Ten sinh vien: {s.get_name()}")
            print(f"GPA: {s.get_gpa()}")
            print(f"Role: {s.get_role()}")
            print("----")
        
```
