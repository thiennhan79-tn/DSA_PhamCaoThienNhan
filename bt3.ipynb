```python
from abc import ABC, abstractmethod
class Person(ABC):
    def __init__(self, id, name):
        self._id = id
        self._name = name

    def get_id(self):
        return self._id
    
    def get_name(self):
        return self._name
    
    def set_id(self, name):
        self._name = name

    @abstractmethod
    def get_role(self):
        pass

    @abstractmethod
    def to_dict(self):
        pass
```

```python
class Student(Person):
    def __init__(self, id, name, gpa):
        super().__init__(id, name)
        self.__gpa = gpa

    def set_gpa(self, gpa):
        self.__gpa = gpa
    
    def get_gpa(self):
        return self.__gpa
    
    def get_role(self):
        return "Student"
    
    def to_dict(self):
        return {
            "id": self._id(),
            "name": self._name(),
            "gpa": self.__gpa()
        }
```

```python
class UnderStudent(Student):
    def __init__(self, id, name, gpa, year):
        super().__init__(id, name, gpa)
        self.__year = year

    def get_year(self):
        return self.__year

    def set_year(self, year):
        self.__year = year

    def get_role(self):
        return "UnderStudent"
    
    def to_dict(self):
        return {
            "id": self._id(),
            "name": self._name(),
            "gpa": self.__gpa(),
            "year": self.__year()
        }
```

```python
class GraduateStudent(Student):
    def __init__(self, id, name, gpa, thesis):
        super().__init__(id, name, gpa)
        self._thesis = thesis

    def get_thesis(self):
        return self._thesis

    def set_thesis(self, thesis):
        self._thesis = thesis

    def get_role(self):
        return "GraduateStudent"
    
    def to_dict(self):
        return {
            "id": self._id(),
            "name": self._name(),
            "gpa": self.__gpa(),
            "thesis": self._thesis()
        }
```
