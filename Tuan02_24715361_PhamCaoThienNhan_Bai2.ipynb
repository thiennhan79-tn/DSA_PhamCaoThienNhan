```python
class HealthcareProfessional:
    def __init__(self, id, name, age, department):
        self.id = id
        self.name = name
        self.age = age
        self.department = department

    def InThongTin(self):
        return f"id:{self.id}, name:{self.name}, age:{self.age}, department:{self.department}"
```

```python
class Doctor(HealthcareProfessional):
    def __init__(self, id, name, age, department, specialization, years_experience):
        super().__init__(id, name, age, department)
        self.specialization = specialization
        self.years_experience = years_experience

    def InThongTin(self):
        return f"Doctor: {super().InThongTin()} - Specialization:{self.specialization} - Experience:{self.years_experience}"
```

```python
class Nurse(HealthcareProfessional):
    def __init__(self, id, name, age, department, certification, shift):
        super().__init__(id, name, age, department)
        self.certification = certification
        self.shift = shift

    def InThongTin(self):
        return f"Nurse: {super().InThongTin()} - Certification:{self.certification} - Shift:{self.shift}"
```

```python
class Stack:
    def __init__(self):
        self.data = []

    def push(self, item):
        self.data.append(item)

    def pop(self):
        if self.isEmpty():
            return None
        return self.data.pop()

    def isEmpty(self):
        return len(self.data) == 0

    def show(self):
        return self.data
```

```python
class HealthcareManagement:
    def __init__(self):
        self.stack = Stack()

    def add_professional(self, p):
        self.stack.push(p)

    def delete(self, id):
        for p in self.stack.data:
            if p.id == id:
                self.stack.data.remove(p)
                return "Da xoa"
        return "Khong tim thay"

    def display(self):
        if self.stack.isEmpty():
            return "Danh sach rong"
        return "\n".join(p.InThongTin() for p in self.stack.show())

    def search_by_id(self, id):
        for p in self.stack.data:
            if p.id == id:
                return p
        return None

    def update(self, id, name=None, age=None, department=None):
        p = self.search_by_id(id)
        if p:
            if name: p.name = name
            if age: p.age = age
            if department: p.department = department
            return "Cap nhat thanh cong"
        return "Khong tim thay"

    def most_experienced_doctor(self):
        doctors = [p for p in self.stack.data if isinstance(p, Doctor)]
        return max(doctors, key=lambda d: d.years_experience)

    def find_oldest(self):
        return max(self.stack.data, key=lambda p: p.age)
    
    def bubble_sort_doctor_experience(self):
        n = len(self.stack.data)
        for i in range (n):
            for j in range (0, n-1-i):
                if self.stack.data[j].years_experience > self.stack.data[j+1].years_experience:
                    self.stack.data[j], self.stack.data[j+1] = self.stack.data[j+1], self.stack.data[j]

    def selection_sort_by_id(self): 
        n = len(self.stack.data)

        for i in range(n):
            min_idx = i
            for j in range(i+1, n):
                if self.stack.data[j].id < self.stack.data[min_idx].id:
                    min_idx = j
            self.stack.data[i], self.stack.data[min_idx] = self.stack.data[min_idx], self.stack.data[i]
```

```python
def menu():
    hm = HealthcareManagement()

    while True:
        print("\n===== HEALTHCARE MANAGEMENT SYSTEM =====")
        print("1. Add healthcare professional (Regular / Doctor / Nurse)")
        print("2. Delete professional")
        print("3. Display professional list")
        print("4. Search professional by ID")
        print("5. Update professional information")
        print("6. Find the most experienced doctor")
        print("7. Find the oldest healthcare professional")
        print("8. Sort doctors by years of experience (Bubble Sort)")
        print("9. Sort professionals by ID (Selection Sort)")
        print("0. Exit")

        choice = input("Choose an option: ")

        if choice == "1":
            print("\n--- Add Professional ---")
            print("R - Regular | D - Doctor | N - Nurse")
            ptype = input("Choose type: ").upper()

            id = input("ID: ")
            name = input("Name: ")
            age = int(input("Age: "))
            department = input("Department: ")

            if ptype == "R":
                p = HealthcareProfessional(id, name, age, department)

            elif ptype == "D":
                spec = input("Specialization: ")
                exp = int(input("Years experience: "))
                p = Doctor(id, name, age, department, spec, exp)

            elif ptype == "N":
                cert = input("Certification: ")
                shift = input("Shift: ")
                p = Nurse(id, name, age, department, cert, shift)

            else:
                print("Invalid type")
                continue

            hm.add_professional(p)
            print("Added successfully")

        elif choice == "2":
            id = input("Enter ID to delete: ")
            print(hm.delete(id))

        elif choice == "3":
            print("\n--- Professional List ---")
            print(hm.display())

        elif choice == "4":
            id = input("Enter ID to search: ")
            p = hm.search_by_id(id)
            print(p.InThongTin() if p else "Not found")

        elif choice == "5":
            id = input("Enter ID to update: ")
            name = input("New name (skip = Enter): ")
            age = input("New age (skip = Enter): ")
            dept = input("New department (skip = Enter): ")

            age = int(age) if age else None
            print(hm.update(id, name or None, age, dept or None))

        elif choice == "6":
            d = hm.most_experienced_doctor()
            print("\nMost experienced doctor:")
            print(d.InThongTin())

        elif choice == "7":
            p = hm.find_oldest()
            print("\nOldest healthcare professional:")
            print(p.InThongTin())

        elif choice == "8":
            hm.bubble_sort_doctor_experience()
            print("Doctors sorted by experience (Bubble Sort)")
            print(hm.display())

        elif choice == "9":
            hm.selection_sort_by_id()
            print("Professionals sorted by ID (Selection Sort)")
            print(hm.display())

        elif choice == "0":
            print("Exit program")
            break

        else:
            print("Invalid choice")
```
