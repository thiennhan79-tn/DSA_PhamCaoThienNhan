```python
class SinhVien:
    def __init__(self, id, name, age = None, gpa = None):
        self.id = id
        self.name = name
        self.age = age
        self.gpa = gpa

    def InThongTin(self):
        return f"id:{self.id}, name:{self.name}, age:{self.age}, gpa:{self.gpa}"
```

```python
class SinhVienChinhQuy(SinhVien):
    def __init__(self, id, name, age=None, gpa=None, noidaotao=None):
        super().__init__(id, name, age, gpa)
        self.noidaotao = noidaotao
    
    def InThongTIn(self):
        return f"{super().InThongTin() } - noi dao tao: {self.noidaotao}"
```

```python
class SinhVienTaiChuc(SinhVien):
    def __init__(self, id, name, age=None, gpa=None, chuyennganh = None):
        super().__init__(id, name, age, gpa)
        self.chuyennganh = chuyennganh

    def InThongTin(self):
        return f"{ super().InThongTin() } - chuyen nganh: {self.chuyennganh}"
```

```python
from abc import ABC, abstractmethod

class BaseSV(ABC):
    
    @abstractmethod
    def isEmpty(self):
        pass

    @abstractmethod
    def push(self, sv):
        pass

    @abstractmethod
    def pop(self):
        pass   
```

```python
class StackSV(BaseSV):
    def __init__(self):
        self.lst_stack = []

    def isEmpty(self):
        return len(self.lst_stack) == 0
    
    def push(self, sv):
        self.lst_stack.append(sv)

    def pop(self):
        #if self.isEmpty() :
         #   return None
        #return self.lst_stack.pop()
        return None if self.isEmpty() else self.lst_stack.pop()
```

```python
class QueueSV(BaseSV):
    def __init__(self):
        self.lst_queue = []

    def isEmpty(self):
        return len(self.lst_queue) == 0
    
    def push(self, sv):
        self.lst_queue.append(sv)

    def pop(self):
        #if self.isEmpty() :
         #   return None
        #return self.lst_queue.pop()
        return None if self.isEmpty() else self.lst_queue.pop(0)
```

```python
class Manage:
    def __init__(self):
        self.ds = StackSV()

    def addSV(self, sv):
        return self.ds.push(sv)
    
    def displaySV(self):
        if self.ds.isEmpty():
            return "Danh sach rong"
        
        return "\n".join((sv.InThongTin()) for sv in self.ds.lst_stack)
        
    def delete_all(self):
        return self.ds.pop()
    
    def delete(self, id):
        found = False
        for sv in self.ds:
            if sv.id == id:
                self.ds.pop()
                found = True  
        return found
    
    def findSV(self, id):
        for sv in self.ds:
            if sv.id == id:
                return sv
        return None
    
    def updateSV(self, id, newSV):
        for sv in self.ds:
            if sv.id == id:
                self.name = newSV.name
                self.age = newSV.age
                self.gpa = newSV.gpa
                return True
        return False
```
