```python
class NhanVien:
    def __init__(self, ma, ho_ten, tuoi=None, luong=None):
        self.ma = ma
        self.ho_ten = ho_ten
        self.tuoi = tuoi
        self.luong = luong

    def __str__(self):
        return f"Ma: {self.ma}, Ho ten: {self.ho_ten}, Tuoi: {self.tuoi}, Luong: {self.luong}"
        
```

```python
class NhanVienVanPhong(NhanVien):
    def __init__(self, ma, ho_ten, tuoi=None, luong=None, phong_ban = None, chuc_vu = None):
        super().__init__(ma, ho_ten, tuoi, luong)
        self.phong_ban = phong_ban
        self.chuc_vu = chuc_vu

    def __str__(self):
        return f"{super.__str__()}, Phong ban: {self.phong_ban}, Chuc vu: {self.chuc_vu}"
        
```

```python
class NhanVienKyThuat(NhanVien):
    def __init__(self, ma, ho_ten, tuoi=None, luong=None, chuyen_mon = None, bac = None):
        super().__init__(ma, ho_ten, tuoi, luong)
        self.chuyen_mon = chuyen_mon
        self.bac = bac

    def __str__(self):
        return f"{super.__str__()}, Chuyen mon: {self.chuyen_mon}, Bac: {self.bac}"
        
```

```python
from abc import ABC, abstractmethod

class Base(ABC):

    @abstractmethod
    def push(self, nv):
        pass

    @abstractmethod
    def isEmpty():
        pass

    @abstractmethod
    def pop(self):
        pass
```

```python
class StackEm(Base):
    def __init__(self):
        self.dsStack = []

    def push(self, nv):
        self.dsStack.append(nv)
    
    def isEmpty(self):
        return len(self.dsStack) == 0
    
    def pop(self):
        #if self.isEmpty() :
         #   return None
        #return self.dsStack.pop()
        return None if self.isEmpty() else self.dsStack.pop()
```

EmployManagement: cung cấp các chức năng quản lý danh sách nhân viên

--1. thêm nhân viên vào danh sách:

    a. Nhập số lượng n, cho người chọn 1. Nhân viên, 2. Nhân viên văn phòng, 3. Nhân viên kỹ thuật

    b. thêm data từ file vào danh sách (txt, json, csv)

--2. Xóa nhân viên

--3. Hiển thị danh sách nhân viên

--4. Tìm kiếm theo mã

--5. Sắp xếp nhân viên theo thuật toán:

    + bubble sort -- lương

    + selection sort -- mã

    +insertion sort -- họ tên

    + merge sort -- tuổi

    + quick sort -- lương hoặc mã

```python
class EmployManagement:
    def __init__(self):
        self.dsNV = []

    def ThemNV(self, nv):
        self.dsNV.append(nv)

    def Them(self, soluong):
        for i in range(soluong):
            loai = int(input("Nhập loại (1. Nhân viên, 2. NVVP, 3. NVKT)"))
            ma = input("Nhập mã")
            ho_ten = input("Nhập họ tên")
            tuoi = input("Nhập tuổi")
            luong = input("Nhập lương")
            if loai == 2:
                phong_ban = input("Nhập phòng ban")
                chuc_vu = input("Nhập chức vụ")
                self.dsNV.append(NhanVienVanPhong(ma, ho_ten, tuoi, luong, phong_ban, chuc_vu))
            elif loai == 3:
                chuyen_mon = input("Nhập chuyên môn")
                bac = input("Nhập bậc")
                self.dsNV.append(NhanVienKyThuat(ma, ho_ten, tuoi, luong, chuyen_mon, bac))
            else:
                self.dsNV.append(NhanVien(ma, ho_ten, tuoi, luong))

    def HienThi(self):
        for nv in self.dsNV:
            print(nv)

    def BubbleSort(self):
        arr = self.dsNV
        n = len(arr)
        for i in range(n):
            for j in range(0, n-1-i):
                if arr[j].luong > arr[j+1].luong:
                    arr[j], arr[j+1] = arr[j+1], arr[j]

    def insertion_sort(self):
        arr = self.dsNV
        n = len(arr)
        for i in range (1, n):
            j = i-1
            while j >= 0 and arr[i].luong < arr[j].luong:
                arr[j+1] = arr[j]
                j -=1
            arr[j+1] = arr[i]
    
    def selection_sort(self):
        arr = self.dsNV
        n = len(arr)
        for i in range(n):
            min_index = i
            for j in range(i+1, n):
                if arr[min_index].ma > arr[j].ma:
                    min_index = j
            arr[i], arr[min_index] = arr[min_index], arr[i]

    def quick_sort(self):
        arr = self.dsNV
        if len(arr) <= 1:
            return arr

        pivot = arr[len(arr) // 2]

        left = [x for x in arr if x.ma < pivot.ma]
        mid = [x for x in arr if x.ma == pivot.ma]
        right = [x for x in arr if x.ma > pivot.ma]

        return self.quick_sort(left) + mid + self.quick_sort(right)
    
    def merge_sort(self):
        arr = self.dsNV
        if len(arr) <= 1:
            return arr
        
        mid = len(arr)//2
        left = arr[:mid]
        right = arr[mid:]

        left_sort = merge_sort(left)
        right_sort = merge_sort(right)

        merge_sort = merge(left_sort, right_sort)

        return merge_sort
    
    def merge(left_sort, right_sort):
        sorted = []
        i = j = 0

        while i < len(left_sort) and j < len(right_sort):
            if left_sort[i] < right_sort[j]:
                sorted.append(left_sort[i])
                i += 1
            else:
                sorted.append(right_sort[j])
                j += 1
        
        while i < len(left_sort):
            sorted.append(left_sort[i])
            i += 1
        while j < len(right_sort):
            sorted.append(right_sort[j])
            j += 1
    
    def doc_file(self, path):
        with open(path, encoding = "utf-8") as f:
            rows = json.load(f)
```
